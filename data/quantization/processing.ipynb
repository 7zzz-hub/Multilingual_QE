```python
languages = ["ar","ca","el","en","es","fr","he","hu","ja","ko","nl","tr","zh"]
len(languages)
```

```python
import json, glob, pandas as pd
from tqdm.notebook import tqdm

records = []
for lang in tqdm(languages):
    files = glob.glob(f'/root/autodl-tmp/data/quantization/wikimedia/20231101.{lang}/*.parquet')
    if files:
        # 只读取第一个文件
        df = pd.read_parquet(files[0])
        # 抽取10条
        sampled_df = df.sample(n=min(1000, len(df)), random_state=42)
        
        for idx, row in sampled_df.iterrows():
            records.append({
                "sample_id": f"{lang}_{idx}",
                "language": lang,
                "source": "wikimedia_20231101",
                "title": row.get("title", ""),
                "text": row["text"]
            })

# 写入文件
with open("calibration_wikipedia_multilingual.jsonl", "w", encoding="utf-8") as f:
    f.write("\n".join([json.dumps(r, ensure_ascii=False) for r in records]))

print(f"总共从 {len(languages)} 个语言中抽取了 {len(records)} 条数据")
```
