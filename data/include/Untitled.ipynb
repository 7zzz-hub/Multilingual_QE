```python
import pandas as pd

languages = ["ar","el","es","fr","he","hu","ja","ko","nl","tr","zh"]

for lang in languages:
    input_path = f"/root/autodl-tmp/Multilingual_QE/data/include/{lang}/test-00000-of-00001.parquet"
    output_path = f"/root/autodl-tmp/Multilingual_QE/data/include/{lang}.json"
    
    df = pd.read_parquet(input_path)
    
    KEEP_KEYS = [
        "language",
        "domain",
        "subject",
        "regional_feature",
        "question",
        "option_a",
        "option_b",
        "option_c",
        "option_d",
        "answer"
    ]
    
    df = df[KEEP_KEYS]
    
    df.to_json(
        output_path,
        orient="records",
        force_ascii=False,
        indent=2
    )

    print("Number of samples:", len(df))
```
