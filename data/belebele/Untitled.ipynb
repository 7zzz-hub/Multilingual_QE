```python
import pandas as pd

languages = ["ar","el","en","ca","es","fr","he","hu","ja","ko","nl","tr","zh"]

for lang in languages:
    input_path = f"/root/autodl-tmp/data/belebele/{lang}.json"
    output_path = f"/root/autodl-tmp/data/belebele/{lang}.json"
    
    df = pd.read_json(input_path)
    
    KEEP_KEYS = [
        "question_number",
        "flores_passage",
        "question",
        "mc_answer1",
        "mc_answer2",
        "mc_answer3",
        "mc_answer4",
        "correct_answer_num"
    ]
    
    df = df[KEEP_KEYS]
    
    df.to_json(
        output_path,
        orient="records",
        force_ascii=False,
        indent=2
    )

    print("Number of samples:", len(df))
```
