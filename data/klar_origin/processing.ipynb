```python
languages = ["ar","ca","el","en","es","fr","he","hu","ja","ko","nl","tr","zh"]
len(languages)
```

### 去除有互相提示的 samples

```python
import glob, json

tmp = []
for lang in languages:
    dataset = []
    files = glob.glob(f"{lang}/*")
    cnt = 0
    for file in files:
        with open(file, 'r', encoding='utf-8') as f:
            data = json.load(f)

        for s in data['samples']:
            if s['subject'] not in s['object'] and s['object'] not in s['subject']:
                tmp.append(s['index'])
```

```python
from collections import Counter

tmp_cnt = Counter(tmp)
idx_done = []
for idx in tmp_cnt:
    if tmp_cnt[idx] == 13:
        idx_done.append(idx)
len(idx_done)
```

```python
import glob, json

for lang in languages:
    dataset = []
    files = glob.glob(f"{lang}/*")
    cnt = 0
    for file in files:
        with open(file, 'r', encoding='utf-8') as f:
            data = json.load(f)

        ### 去除跟语言相关的内容
        if "language" in data['name']:
            continue
        
        tmp = []
        for s in data['samples']:
            if s['index'] in idx_done:
                tmp.append(s)
                
        dataset.append({
            'relation': data['name'],
            'prompt_templates': data['prompt_templates'],
            'samples': tmp
        })

        cnt += len(tmp)
    print(lang, cnt)
    
    with open(f"../klar/{lang}.json", 'w', encoding='utf-8') as f:
        json.dump(dataset, f, indent=2)
```
